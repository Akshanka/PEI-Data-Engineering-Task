```python
try:
    from src.utils.common import (
        read_config,
        build_table_name,
        write_delta_table
    )

    from src.transformations.customers import (
        transform_enriched_customers
    )

    from src.transformations.products import (
        transform_enriched_products
    )

    print("Imports completed successfully")

except Exception as error:
    raise RuntimeError(f"Failed during imports: {error}")
```

```python
try:
    config = read_config("../config/dev_config.json")

    print("Config loaded successfully")

except Exception as error:
    raise RuntimeError(f"Failed to read config: {error}")
```

```python
try:
    refined_customers_table = build_table_name(
        config,
        "refined",
        "customers"
    )

    refined_products_table = build_table_name(
        config,
        "refined",
        "products"
    )

    enriched_customers_table = build_table_name(
        config,
        "enriched",
        "customers"
    )

    enriched_products_table = build_table_name(
        config,
        "enriched",
        "products"
    )

    print("Table names created successfully")

except Exception as error:
    raise RuntimeError(f"Failed to build table names: {error}")
```

```python
try:
    refined_customers_df = spark.table(
        refined_customers_table
    )

    refined_products_df = spark.table(
        refined_products_table
    )

    print("Refined tables read successfully")

except Exception as error:
    raise RuntimeError(f"Failed to read refined tables: {error}")
```

```python
try:
    enriched_customers_df = transform_enriched_customers(
        refined_customers_df
    )

    enriched_products_df = transform_enriched_products(
        refined_products_df
    )

    print("Enriched transformations completed successfully")

except Exception as error:
    raise RuntimeError(
        f"Failed during enriched transformations: {error}"
    )
```

```python
enriched_customers_df.printSchema()
```

```python
try:
    write_delta_table(
        enriched_customers_df,
        enriched_customers_table,
        mode="overwrite"
    )

    write_delta_table(
        enriched_products_df,
        enriched_products_table,
        mode="overwrite"
    )

    print("Enriched tables written successfully")

except Exception as error:
    raise RuntimeError(
        f"Failed to write enriched tables: {error}"
    )
```
