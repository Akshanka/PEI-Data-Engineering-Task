```python
try:
    from src.utils.common import (
        read_config,
        build_table_name,
        write_delta_table
    )

    from src.transformations.orders import (
        transform_enriched_orders
    )

    print("Imports completed successfully")

except Exception as error:
    raise RuntimeError(f"Failed during imports: {error}")
```

```python
try:
    config = read_config("../config/dev_config.json")

    print("Config loaded successfully")

except Exception as error:
    raise RuntimeError(f"Failed to read config: {error}")
```

```python
try:
    refined_orders_table = build_table_name(
        config,
        "refined",
        "orders"
    )

    enriched_customers_table = build_table_name(
        config,
        "enriched",
        "customers"
    )

    enriched_products_table = build_table_name(
        config,
        "enriched",
        "products"
    )

    enriched_orders_table = build_table_name(
        config,
        "enriched",
        "orders"
    )

    print("Table names created successfully")

except Exception as error:
    raise RuntimeError(f"Failed to build table names: {error}")
```

```python
try:
    refined_orders_df = spark.table(
        refined_orders_table
    )

    enriched_customers_df = spark.table(
        enriched_customers_table
    )

    enriched_products_df = spark.table(
        enriched_products_table
    )

    print("Tables read successfully")

except Exception as error:
    raise RuntimeError(f"Failed to read tables: {error}")
```

```python
try:
    enriched_orders_df = transform_enriched_orders(
        refined_orders_df,
        enriched_customers_df,
        enriched_products_df
    )

    print("Enriched orders transformation completed successfully")

except Exception as error:
    raise RuntimeError(
        f"Failed during enriched orders transformation: {error}"
    )
```

```python
enriched_orders_df.printSchema()
```

```python
try:
    write_delta_table(
        enriched_orders_df,
        enriched_orders_table,
        mode="overwrite"
    )

    print("Enriched orders table written successfully")

except Exception as error:
    raise RuntimeError(
        f"Failed to write enriched orders table: {error}"
    )
```

```python
try:
    print("Refined Orders Count")
    print(
        refined_orders_df.count()
    )

    print("Enriched Orders Count")
    print(
        spark.table(enriched_orders_table).count()
    )

except Exception as error:
    raise RuntimeError(
        f"Failed during validation: {error}"
    )
```

```python
try:
    spark.table(
        enriched_orders_table
    ).printSchema()

    display(
        spark.table(enriched_orders_table)
    )

except Exception as error:
    raise RuntimeError(
        f"Failed while checking enriched orders output: {error}"
    )
```
