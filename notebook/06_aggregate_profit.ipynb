```python
try:
    from src.utils.common import read_config, build_table_name, write_delta_table
    from src.transformations.aggregations import transform_profit_aggregate

    print("Imports completed successfully")

except Exception as error:
    raise RuntimeError(f"Failed during imports: {error}")
```

```python
try:
    config = read_config("../config/dev_config.json")

    print("Config loaded successfully")

except Exception as error:
    raise RuntimeError(f"Failed to read config: {error}")
```

```python
try:
    enriched_orders_table = build_table_name(config, "enriched", "orders")

    aggregate_profit_table = build_table_name(
        config,
        "aggregate",
        "profit_by_year_category_subcategory_customer"
    )

    print("Table names created successfully")

except Exception as error:
    raise RuntimeError(f"Failed to build table names: {error}")
```

```python
try:
    enriched_orders_df = spark.table(enriched_orders_table)

    print("Enriched orders table read successfully")

except Exception as error:
    raise RuntimeError(f"Failed to read enriched orders table: {error}")
```

```python
try:
    aggregate_profit_df = transform_profit_aggregate(enriched_orders_df)

    print("Aggregate transformation completed successfully")

except Exception as error:
    raise RuntimeError(f"Failed during aggregate transformation: {error}")
```

```python
try:
    write_delta_table(
        aggregate_profit_df,
        aggregate_profit_table,
        mode="overwrite"
    )

    print("Aggregate table written successfully")

except Exception as error:
    raise RuntimeError(f"Failed to write aggregate table: {error}")
```

```python
try:
    spark.table(aggregate_profit_table).printSchema()
    display(spark.table(aggregate_profit_table))

except Exception as error:
    raise RuntimeError(f"Failed during aggregate validation: {error}")
```
